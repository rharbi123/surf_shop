```python
import pandas as pd
import json

with open('../data/raw/meteo.json', 'r', encoding='utf-8') as f:
    meteo = json.load(f)

# Aplatir les données (spots + prévisions)
rows = []
for spot in meteo:
    for prev in spot['previsions']:
        row = {'spot': spot['spot'], 'ville': spot['ville'], 'pays': spot['pays']}
        row.update(prev)
        rows.append(row)

df = pd.DataFrame(rows)
print(f"Total lignes: {len(df)}")
df.head()
```

```python
print(df.info())
print("\nDonnées manquantes:")
print(df.isnull().sum())

print(f"Doublons: {df.duplicated().sum()}")
```
