```python
import pandas as pd
import json

# Charger le JSON
with open('../data/raw/planches.json', 'r') as f:
    planches = json.load(f)

with open('../data/raw/planches.json', 'r', encoding='utf-8') as f:
    planches = json.load(f)

df = pd.DataFrame(planches)
print(f"Total planches: {len(df)}")
df.head()
```

```python
# Infos sur le dataset
print(df.info())
print("\nDonnées manquantes:")
print(df.isnull().sum())
```

```python
print(f"Doublons: {df.duplicated().sum()}")
```

```python
print("Exemples de chaque colonne:")
for col in df.columns:
    print(f"\n{col}:")
    print(df[col].iloc[:3].values)
```

```python
# Chercher les "Non disponible"
for col in df.columns:
    count = (df[col] == "Non disponible").sum()
    if count > 0:
        print(f"{col}: {count}")
```
